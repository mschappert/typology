```python
#chunk = {'band': 1, 'x': 1024, 'y': 1024}

def img_diff(image1, image2, output_dir, metric = metric): # chunks=chunk
    """
    Subtract two TIFF rasters using rioxarray + Dask, preserving metadata
    and allowing negative values.

    Parameters:
    image1_path (str): Path to the earlier image (e.g., 1990).
    image2_path (str): Path to the later image (e.g., 1995).
    output_dir (str): Path where the difference raster will be saved.
    chunks (dict): Dask chunk sizes by dimension.
    """
    
    # Open each raster lazily with Dask
    img1 = rxr.open_rasterio(image1, chunks= 'auto')#{'band': 1, 'x': 512, 'y': 512}) # chunks=chunks
    img2 = rxr.open_rasterio(image2, chunks= 'auto')#{'band': 1, 'x': 512, 'y': 512}) # chunks=chunks

    # Align dimensions and coordinates if needed
    #img2, img1 = xr.align(img2, img1, join='exact')

    # subtract images (later minus earlier)
    diff = img2 - img1

    # Preserve CRS and transform, then write with Dask in parallel
    diff.rio.write_crs(img1.rio.crs, inplace=True)
    diff.rio.write_transform(img1.rio.transform(), inplace=True)
    
    # create naming convention
    base1 = os.path.splitext(os.path.basename(image1))[0]
    base2 = os.path.splitext(os.path.basename(image2))[0]
    # pull year from begining of name (ie 1990)
    year1 = ''.join(filter(str.isdigit, base1))[:4]
    year2 = ''.join(filter(str.isdigit, base2))[:4]
    # take just last 2 digits (ie 90)
    label1 = year1[-2:]
    label2 = year2[-2:]
    
    # output file name "90-95_area.tif"
    out_name = f"{label1}-{label2}_{metric}.tif"
    out_path = os.path.join(output_dir, out_name)
    
    #diff_computed = diff.compute() 

    # Write raster using Dask
    diff.rio.to_raster(
        out_path,
        dtype='int32',
        compress='LZW',
        tiled=True #,
        #num_workers='all_cpus',  # use all available cores
        #lock=True
    )

    return output_dir
```

```python
# Simple rasterio function
def img_diff(image1, image2, output_dir, metric):
    import os
    import rasterio
    
    os.makedirs(output_dir, exist_ok=True)
    
    with rasterio.open(image1) as src1, rasterio.open(image2) as src2:
        diff = src2.read(1).astype('float32') - src1.read(1).astype('float32')
        
        profile = src1.profile.copy()
        profile.update(dtype='float32', compress='lzw')
        
        year1 = os.path.basename(image1)[:4]
        year2 = os.path.basename(image2)[:4]
        output_file = os.path.join(output_dir, f"{year1}-{year2}_{metric}.tif")
        
        with rasterio.open(output_file, 'w', **profile) as dst:
            dst.write(diff, 1)
    
    return output_file

# Run it
result = img_diff(earlier, later, output, metric)
print(f"Saved: {result}")
```

```python
# Close current cluster
# client.close()
# cluster.close()

# Create smaller cluster
from dask.distributed import Client, LocalCluster
cluster = LocalCluster(n_workers=2, threads_per_worker=2, memory_limit='12GB')
client = Client(cluster)
print(f"Dashboard: {client.dashboard_link}")                                                                                                                                                                                                                                                                      

def img_diff(image1, image2, output_dir, metric):
    import os
    import rioxarray as rxr
    
    os.makedirs(output_dir, exist_ok=True)
    
    # Very small chunks
    img1 = rxr.open_rasterio(image1, chunks={'x': 256, 'y': 256})
    img2 = rxr.open_rasterio(image2, chunks={'x': 256, 'y': 256})
    
    diff = img2 - img1
    
    year1 = os.path.basename(image1)[:4]
    year2 = os.path.basename(image2)[:4]
    output_file = os.path.join(output_dir, f"{year1}-{year2}_{metric}.tif")
    
    # Write with compute=False to avoid memory spike
    diff.rio.to_raster(output_file, dtype='float32', compress='lzw', compute=True)
    
    return output_file

result = img_diff(earlier, later, output, metric)
print(f"Saved: {result}")
```
